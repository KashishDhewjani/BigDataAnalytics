Pandas: A Very Basic Introduction
=================================

`pandas` is a Python library used to work with data.

In the previous notebook we used pandas to put information returned by the ONS API into a table so that it was easier to view.

In this notebook we will only look at a few basic pandas features:

- importing pandas
- creating a DataFrame
- viewing data
- checking rows and columns
- selecting columns
- selecting simple rows

There is **no ETL in this notebook**.

Import pandas
-------------

The usual way to import pandas is:

```python
import pandas as pd
```

`pd` is simply a short name for pandas.

We can now use pandas commands by writing `pd.` followed by the command we need.

A DataFrame
-----------

A **DataFrame** is a table made of rows and columns.

This is the main pandas object we used in the ONS data acquisition notebook.

We can create a small DataFrame from Python data:

```python
data = [
    {"area": "Peterborough", "year": 2021, "value": 82.1},
    {"area": "Cambridge", "year": 2021, "value": 84.3},
    {"area": "Leicester", "year": 2021, "value": 81.7},
    {"area": "Peterborough", "year": 2022, "value": 82.5},
    {"area": "Cambridge", "year": 2022, "value": 84.6}
]

df = pd.DataFrame(data)

df
```

The variable `df` now contains a pandas DataFrame.

Each dictionary became one row.

The dictionary keys became the column names.

Viewing the first rows
----------------------

For large datasets we usually do not want to print every row.

`head()` shows the first five rows:

```python
df.head()
```

We can also choose how many rows to display:

```python
df.head(3)
```

This is the same pandas feature used in the earlier ONS notebook when we wrote:

    df.head(10)

Checking the size of the DataFrame
----------------------------------

`shape` tells us how many rows and columns are in the DataFrame.

```python
df.shape
```

The result is shown as:

    (rows, columns)

For example, `(5, 3)` means 5 rows and 3 columns.

Checking the column names
-------------------------

`columns` shows the names of all columns:

```python
df.columns
```

Checking the data types
-----------------------

`dtypes` shows the type of data stored in each column.

```python
df.dtypes
```

For example:

- text is usually stored as `object`
- whole numbers may be stored as `int`
- decimal numbers may be stored as `float`

Selecting one column
--------------------

Use square brackets and the column name:

```python
df["area"]
```

The result is one pandas **Series**.

A Series is similar to one column of a DataFrame.

Selecting more than one column
------------------------------

To select several columns, put the column names inside a list:

```python
df[["area", "value"]]
```

Selecting rows with a simple condition
--------------------------------------

We can ask pandas to show only rows that meet a condition.

For example, show only rows where the area is Peterborough:

```python
df[df["area"] == "Peterborough"]
```

Another example: show rows where the value is greater than 83:

```python
df[df["value"] > 83]
```

Simple summary information
--------------------------

`info()` gives a quick overview of the DataFrame:

```python
df.info()
```

It shows information such as:

- number of rows
- column names
- whether values are present
- data types

For now, the important point is simply that `info()` helps us understand what is inside a DataFrame.
