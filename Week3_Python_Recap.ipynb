Now lets create a class. In Python, every class should derive from object. Our class will describe a person, with a name and an age. We will supply a constructor, and a method to get the full name.

```python
class Person(object):
    def __init__(self, first, last, age):
        self.first = first
        self.last = last
        self.age = age
    
    def full_name(self):
        return self.first + ' ' + self.last
```

```python
class Car(object):
    def __init__(self,model,colour, numberOfWhels):
        self.model = model
        self.colour = colour
        self.numberOfWheels = numberOfWhels
        
    def carDetails(self):
        return  'This car model is ' +self.model + ' ' + self.colour + ' ' + self.numberOfWheels
        
            
```

```python
car = Car('toyota', 'grey', '4')
print(car.carDetails())
```

Now Let us Learn about regex

```python
#In python to use the regular expression module, we need to import it first.
import re

Input = 'The cat sat on the mat.'
pattern = r'cat'
result = re.search(pattern, Input)
print(result)
```

```python
text = """dog
Dog
hog
blog
catalog
scatterdog
dogmatic"""

# 1. Match only the complete line "dog"
print("1:", re.findall(r"^dog$", text, re.MULTILINE))

# 2. Match complete lines "dog" and "Dog"
print("2:", re.findall(r"^dog$", text, re.MULTILINE | re.IGNORECASE))

# 3. Match "dog" at the start of a line, including in "dogmatic"
print("3:", re.findall(r"^dog", text, re.MULTILINE | re.IGNORECASE))
```

```python
import re

lines = [
    "The Matrix",
    "Matrix Reloaded",
    "Animatrix",
    "Enter the Matrix",
    "Matrixed",
    "Matrix123"
]

# Print the full lines containing each match
patterns = {
    "1. Starts with Matrix": r"^Matrix",
    "2. Ends with Matrix": r"Matrix$",
    "3. Contains the exact word Matrix": r"\bMatrix\b"
}

for label, pattern in patterns.items():
    print(label)
    for line in lines:
        if re.search(pattern, line):
            print(" ", line)
```

```python
import re

text = """abc
abbc
abbbc
ac
abbcde
abbbbbbbbbc"""

# 1. Exactly 2 or 3 b's
print("1:", re.findall(r"^ab{2,3}c$", text, re.MULTILINE))

# 2. One or more b's
print("2:", re.findall(r"^ab+c$", text, re.MULTILINE))

# 3. Zero or more b's
print("3:", re.findall(r"^ab*c$", text, re.MULTILINE))
```

```python
import re

text = """Email me at john.doe@example.com
Or at jane99@test.org
Also contact admin@my-site.co.uk
Invalid: test@site
Strange: fake@domain.comm
Typo: user@@example.com"""

# Boundaries prevent extracting part of an invalid email.
# These patterns are for this exercise, not full email validation.

pattern1 = (
    r"(?<![\w.%+@-])"
    r"[A-Za-z0-9._%+-]+@"
    r"[A-Za-z0-9-]+(?:\.[A-Za-z0-9-]+)*"
    r"\.(?:com|org)"
    r"(?![\w@-]|\.[A-Za-z0-9])"
)

pattern2 = (
    r"(?<![\w.%+@-])"
    r"[A-Za-z0-9._%+-]+@"
    r"[A-Za-z0-9-]+(?:\.[A-Za-z0-9-]+)*"
    r"\.(?:com|org|co\.uk)"
    r"(?![\w@-]|\.[A-Za-z0-9])"
)

print("1:", re.findall(pattern1, text))
print("2:", re.findall(pattern2, text))
```

```python
import re

text = """Order #12345 placed on 2025-10-01
Order #54321 placed on 2025-09-15
Order #88888 placed on 2025-08-20
Reference 99999 on 2025-10-01
Order #123 placed on Oct 1st 2025"""

# 1. Extract numbers preceded by "Order #"
order_numbers = re.findall(r"Order #(\d+)\b", text)
print("1. Order numbers:", order_numbers)

# 2. Extract all dates in YYYY-MM-DD format
dates = re.findall(r"\b(\d{4}-\d{2}-\d{2})\b", text)
print("2. Dates:", dates)

# 3. Extract order number and date together
pattern = r"Order #(\d+) placed on (\d{4}-\d{2}-\d{2})\b"
matches = re.findall(pattern, text)

print("3. Orders with dates:")
for order_number, date in matches:
    print(f"Order: {order_number}, Date: {date}")
```

```python
#Matching and extracting the group from the characters that match the pattern.
#Matching the price of the items in the order list.

ListOfItems = [
    "Order #101: 2x Apple @ $1.50",
    "Order #102: 1x Banana @ $0.75",
    "Order #103: 3x Orange @ $2.00",
    "$1.50",
    " something else$0.75",
    "$2.00 don't want it"
]

#\d+x\s+(?:Apple|Banana|Orange)\s+@\s+\$(\d+\.\d{2})

pattern = r"\d+x\s+(?:Apple|Banana|Orange)\s+@\s+\$(\d+\.\d{2})"

for item in ListOfItems:
    match = re.search(pattern, item)
    if match:
        print(match.group(1))  # Extract the captured price
```
